```python
import torch
import core.Environment as e
import core.Paramecium as p

from typing import Literal
```

```python
class PLModule:
    COORD = 0
    SURROUND_VALUES = 1
    FOOD_VALUES = 2
    def __init__(self, paramecia:dict):
        self.paramecia = paramecia  #1:[(y,x), [0,0,0,0,0,0,0,0], 0]

    def __call__(self, choice: Literal["COORD", "SURROUND_VALUES", "FOOD_VALUES"] = "SURROUND_VALUES"):
        """
        通过调用其本身来返回所有（在paramecia字典中注册过的）草履虫的
        |坐标 -->COORD
        |周围的化学信息 -->SURROUND_VALUES
        |吃到的食物 -->FOOD_VALUES
        它们都是torch.tensor类型的数据，用List类型包裹了不同个体的数据
        """
        if choice == "COORD":
            choice = self.COORD
        elif choice == "SURROUND_VALUES":
            choice = self.SURROUND_VALUES
        elif choice == "FOOD_VALUES":
            choice = self.FOOD_VALUES

        return [torch.tensor(self.paramecia[key][choice]) for key in self.paramecia.keys()]

```

```python
environment = e.Environment(food_num = 20)
paramecium = p.Paramecium(environment, p_num = 10)
pl_module = PLModule(paramecium.paramecia)

def setup(_environment: e.Environment):
    environment.generate_food()
    paramecium.generate_paramecium()


def loop(_environment: e.Environment):
    environment.spread()
    paramecium.move_randomly()
    paramecium.sense_value()
    # paramecium()
    # print(torch.tensor(paramecium.paramecia[1][1]))
    print(pl_module("COORD"))

def special(click_y,click_x):
    paramecium.update_click(click_y,click_x)

e.start(environment, setup, loop, special, click_obj = 1, fps = 60)
```
